```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEmbeddings
from sklearn.metrics.pairwise import cosine_similarity
```

```python
documents = [
    "My name is Ajitesh",
    "I am from India",
    "I am a 4th undergraduate student",
    "I love programming",
    "I am in Computer Science and Engineering department at IIT Kharagpur",
]


doc_embeddings =[]
```

```python
for doc in documents:
    embeddings = HuggingFaceEmbeddings(model_name='sentence-transformers/all-MiniLM-L6-v2')
    vector = embeddings.embed_query(doc)
    doc_embeddings.append(vector)
print(doc_embeddings)
```

```python
query = "where do i study?"
query_embedding = embeddings.embed_query(query)
```

```python
# minimum_similarity = -1
# most_similar_doc = ""
# for i in doc_embeddings:
#     similarity = cosine_similarity([query_embedding], [i])
#     print(similarity[0][0])
#     if similarity[0][0] > minimum_similarity:
#         minimum_similarity = similarity[0][0]
#         most_similar_doc = documents[doc_embeddings.index(i)]

#rank the similarities
similarities = []
for i in doc_embeddings:
    similarity = cosine_similarity([query_embedding], [i])
    similarities.append(similarity[0][0])
most_similar_doc = documents[similarities.index(max(similarities))]
print("Most similar document is: ", most_similar_doc)
print("All similarities: ", similarities)
for i in range(len(documents)):
    for j in range(i+1, len(documents)):
        if (similarities[i] < similarities[j]):
            # swap
            documents[i], documents[j] = documents[j], documents[i]
            similarities[i], similarities[j] = similarities[j], similarities[i]

print("Ranked documents: ", documents)

print("\n\n\n\nRanked similarities: ", similarities)
```
